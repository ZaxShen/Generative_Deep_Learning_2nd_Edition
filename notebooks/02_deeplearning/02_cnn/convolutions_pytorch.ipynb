# 🎆 Convolutions

In this notebook, we'll walk through how convolutional filters can pick out different aspects of an image

```python
%matplotlib inline
import matplotlib.pyplot as plt
import numpy as np
from skimage import data
from skimage.color import rgb2gray
from skimage.transform import resize

import torch
import torch.nn.functional as F
```

## 0. Original Input Image <a name="orig"></a>

```python
im = rgb2gray(data.coffee())
im = resize(im, (64, 64))
print(im.shape)

# PyTorch convolutions expect a batch of images shaped (N, C, H, W):
# here 1 image with 1 (grey) channel
im_tensor = torch.tensor(im, dtype=torch.float32)[None, None]
print(im_tensor.shape)

plt.axis("off")
plt.imshow(im, cmap="gray")
```

## Horizontal Edge Filter <a name="hor"></a>

```python
filter1 = torch.tensor([[1, 1, 1], [0, 0, 0], [-1, -1, -1]], dtype=torch.float32)

# F.conv2d slides the filter over the image and, at every position, multiplies
# the 3x3 patch by the filter element-wise and sums the result.
# The filter must be shaped (out_channels, in_channels, kH, kW) = (1, 1, 3, 3);
# padding=1 adds a border of zeros so the output stays 64x64
new_image = F.conv2d(im_tensor, filter1[None, None], padding=1)
new_image = new_image[0, 0]  # (1, 1, H, W) -> (H, W) for plotting

plt.axis("off")
plt.imshow(new_image.numpy(), cmap="Greys")
```

## Vertical Edge Filter <a name="ver"></a>

```python
filter2 = torch.tensor([[-1, 0, 1], [-1, 0, 1], [-1, 0, 1]], dtype=torch.float32)

new_image = F.conv2d(im_tensor, filter2[None, None], padding=1)[0, 0]

plt.axis("off")
plt.imshow(new_image.numpy(), cmap="Greys")
```

## Horizontal Edge Filter with Stride 2 <a name="hor2"></a>

```python
filter1 = torch.tensor([[1, 1, 1], [0, 0, 0], [-1, -1, -1]], dtype=torch.float32)

stride = 2

# With stride=2 the filter jumps 2 pixels at a time, halving the output size
new_image = F.conv2d(im_tensor, filter1[None, None], stride=stride, padding=1)[0, 0]
print(new_image.shape)

plt.axis("off")
plt.imshow(new_image.numpy(), cmap="Greys")
```

## Vertical Edge Filter with Stride 2 <a name="ver2"></a>

```python
filter2 = torch.tensor([[-1, 0, 1], [-1, 0, 1], [-1, 0, 1]], dtype=torch.float32)

stride = 2

new_image = F.conv2d(im_tensor, filter2[None, None], stride=stride, padding=1)[0, 0]
print(new_image.shape)

plt.axis("off")
plt.imshow(new_image.numpy(), cmap="Greys")
```
